```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
datadir = '/staging/leuven/stg_00187/users/Irati/data'
```

```python
WT_pca= []
for rep in range(1, 4):
    df = pd.read_csv(f'{datadir}/1YVL_noNTD/WT/rep{rep}/analysis/003_pca/2dproj_timescale.csv', header=None)
    WT_pca.append(df)

T385M_pca= []
for rep in range(1, 4):
    df = pd.read_csv(f'{datadir}/1YVL_noNTD/T385M/rep{rep}/analysis/003_pca/2dproj_timescale.csv', header=None)
    T385M_pca.append(df)

R274W_T385M_pca= []
for rep in range(1, 4):
    df = pd.read_csv(f'{datadir}/1YVL_noNTD/R274W_T385M/rep{rep}/analysis/003_pca/2dproj_timescale.csv', header=None)
    R274W_T385M_pca.append(df)
```

```python
WT_pca[1].columns
```

```python
all_pca = WT_pca + T385M_pca + R274W_T385M_pca
y_min = min(df[1].min() for df in all_pca)
y_max = max(df[1].max() for df in all_pca)
```

```python
fig, axes = plt.subplots(3, 3, figsize=(16, 12), sharex=True, sharey=True)

# Plot WT replicates
for i, (df, ax) in enumerate(zip(WT_pca, axes[0, :])):
    scatter = ax.scatter(df[0], df[1], c=df[2], cmap='viridis', s=10)
    ax.set_title(f'WT Replicate {i+1}', fontweight='bold')
    ax.set_xlabel('PC1 (Å)')
    ax.set_ylabel('PC2 (Å)')
    ax.set_axisbelow(True)
    ax.grid(True)

# Plot T385M replicates
for i, (df, ax) in enumerate(zip(T385M_pca, axes[1, :])):
    scatter = ax.scatter(df[0], df[1], c=df[2], cmap='viridis', s=10)
    ax.set_title(f'T385M Replicate {i+1}', fontweight='bold')
    ax.set_xlabel('PC1 (Å)')
    ax.set_ylabel('PC2 (Å)')
    ax.set_axisbelow(True)
    ax.grid(True)

# Plot R274W_T385M replicates
for i, (df, ax) in enumerate(zip(R274W_T385M_pca, axes[2, :])):
    scatter = ax.scatter(df[0], df[1], c=df[2], cmap='viridis', s=10)
    ax.set_title(f'R274W_T385M Replicate {i+1}', fontweight='bold')
    ax.set_xlabel('PC1 (Å)')
    ax.set_ylabel('PC2 (Å)')
    ax.set_axisbelow(True)
    ax.grid(True)

# Adjust layout and add shared colorbar to the right
cbar = fig.colorbar(scatter, ax=axes, location='right', shrink=0.2, label='Time (ns)')


plt.subplots_adjust(hspace=0.5, right=0.75)
plt.show()
```
