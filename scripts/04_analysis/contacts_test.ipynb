```python
from MDAnalysis.analysis.contacts import Contacts
from MDAnalysis.lib.distances import distance_array
import MDAnalysis as mda
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
datadir = '../../data/1bf5/wt/rep2/gromacs'
```

```python
u = mda.Universe(f"{datadir}/md.gro", f"{datadir}/md_noPBC_mol_com_cluster_fit.xtc")
sel1 = u.select_atoms("protein or resname PTR")
sel2 = u.select_atoms("nucleic")

# Distance-based contact analysis
cont = Contacts(u, select=(sel1, sel2), refgroup=(sel1, sel2), radius=5.0)
cont.run(start=1800, stop=2001, verbose=True)
```

```python
# Results: fraction of contacts over time
print(cont.results.timeseries)
```

```python
pd.DataFrame(cont.results.timeseries).head()
```

```python
average_contacts = np.mean(cont.results.timeseries[:, 1])
print('average contacts = {}'.format(average_contacts))
```

```python
fig, ax = plt.subplots()
ax.plot(cont.results.timeseries[:, 0], cont.results.timeseries[:, 1])
ax.set(xlabel='frame', ylabel='fraction of native contacts',
       title='Native Contacts, average = {:.2f}'.format(average_contacts))
fig.show()
```

### Distance by residue pairs

```python
# Define selections
sel1 = u.select_atoms("protein and resid 182:353")  # Protein residues 182-353
sel2 = u.select_atoms("nucleic")                    # DNA/RNA

# Parameters
cutoff = 5.0  # Distance cutoff in Å
start_frame = 1800
stop_frame = 2001

# List to store contact data
contact_data = []

# Loop over trajectory
for ts in u.trajectory[start_frame:stop_frame]:
    frame = ts.frame
    # Calculate distance matrix between sel1 and sel2
    dist_matrix = distance_array(sel1.positions, sel2.positions, box=u.dimensions)
    
    # Find pairs within cutoff
    close_pairs = np.where(dist_matrix <= cutoff)
    
    # Collect contact details
    for i, j in zip(close_pairs[0], close_pairs[1]):
        atom1 = sel1[i]
        atom2 = sel2[j]
        contact = {
            "frame": frame,
            "sel1_resid": f"{atom1.resid}", 
            "sel1_resnm": f"{atom1.resname}", 
            "sel2_resid": f"{atom2.resid}", 
            "sel2_resnm": f"{atom2.resname}",
            "distance": dist_matrix[i, j]
        }
        contact_data.append(contact)

# Create DataFrame
contacts_df = pd.DataFrame(contact_data)
```
